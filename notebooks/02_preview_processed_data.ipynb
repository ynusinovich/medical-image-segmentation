## Imports

```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join(os.path.dirname("__file__"), '..')))
import random
import numpy as np
import matplotlib.pyplot as plt
import imageio
```

## Paths


```python
base_dir = "../data/processed/"
splits = ['train', 'val', 'test']
split = "train"  # or "val" or "test"

image_dir = os.path.join(base_dir, split, "images")
mask_dir = os.path.join(base_dir, split, "masks")
```

## List Files

```python
image_files = sorted(os.listdir(image_dir))
mask_files = sorted(os.listdir(mask_dir))

print(f"Found {len(image_files)} images in {split} split")
```

## Pick Random Samples and Plot

```python
def plot_random_samples(num_samples=5):
    selected_indices = random.sample(range(len(image_files)), num_samples)

    fig, axs = plt.subplots(num_samples, 2, figsize=(8, 4 * num_samples))

    for i, idx in enumerate(selected_indices):
        img_path = os.path.join(image_dir, image_files[idx])
        mask_path = os.path.join(mask_dir, mask_files[idx])

        img = imageio.imread(img_path)
        mask = imageio.imread(mask_path)

        axs[i, 0].imshow(img, cmap="gray")
        axs[i, 0].set_title(f"Image: {image_files[idx]}")
        axs[i, 0].axis('off')

        axs[i, 1].imshow(mask, cmap="gray", vmin=0, vmax=2)
        axs[i, 1].set_title(f"Mask: {mask_files[idx]}")
        axs[i, 1].axis('off')

    plt.tight_layout()
    plt.show()

plot_random_samples(5)
```
